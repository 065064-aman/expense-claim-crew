```python
# Load the Gemini key from the .env file (kept out of GitHub)
import os
from dotenv import load_dotenv

load_dotenv()
key = os.getenv("GEMINI_API_KEY")
print("Key found:", bool(key), "| starts with:", key[:4] if key else None)
```

```python
import crewai
from crewai import Agent, Task, Crew, LLM

print("CrewAI", crewai.__version__)
llm = LLM(model="gemini/gemini-flash-latest", api_key=key, temperature=0.2)
```

```python
# One agent, one task: the smallest possible crew
auditor = Agent(
    role="Expense Policy Expert",
    goal="Check employee expense claims against company policy",
    backstory="You are a senior finance auditor at an Indian company with 15 years of experience in expense reimbursements.",
    llm=llm,
    verbose=False,
)

check = Task(
    description="An employee claims Rs 4,500 for a client dinner. The policy limit for client meals is Rs 3,000 per person. Is this within policy? Answer in two sentences.",
    expected_output="A two-sentence verdict.",
    agent=auditor,
)

result = Crew(agents=[auditor], tasks=[check], verbose=False).kickoff()
print(result)
```
